```python
# We need to convert datetime to numeric for training.
# Let's extract key features into separate numeric columns
def add_features(df):
    df['month'] = pd.to_datetime(df['datetime']).dt.month
    df['Year'] = pd.to_datetime(df['datetime']).dt.year
    df['Day'] = pd.to_datetime(df['datetime']).dt.day
    df['Hour'] = pd.to_datetime(df['datetime']).dt.hour
    return df
```

```python
df = add_features(df)
#add_features(df_test)
df.head(3)
```

```python
data = {
    "Age": [22, 25, 28, 32, 35],
    "Experience": [1, 3, 5, 8, 10],
    "Salary": [40000, 48000, 55000, 68000, 75000]
}

data
```

```python
import pandas as pd

df = pd.DataFrame(data)
df
```

```python
def pay_status(salary, experience):
    expected_salary = 35000 + (experience * 5000)

    if salary >= expected_salary:
        return "Highly Paid"
    else:
        return "Under Paid"
```

```python
df["Pay_Status"] = df.apply(pay_status, axis=1)
```

```python
df["Pay_Status"] = df.apply(
    lambda row: pay_status(row["Salary"], row["Experience"]),
    axis=1
)
```

```python
def salary_status(salary):
    if salary >= 60000:
        return "High Salary"
    else:
        return "Low Salary"

df["Salary_Status"] = df["Salary"].map(salary_status)
```
